### Gradients


```python
import numpy as np
import matplotlib.pyplot as plt

# define cost function

def cost_function(x):
    return x**2

# define derivative
def gradient(x):
    return x*2


# Gradient Descent Paramenter

learning_rate = 0.8 # How big a step to take
initial_x = 10.0 #starting point of X
num_iterations = 50 #how many steps to make


x_history = [initial_x]

cost_history = [cost_function(initial_x)]

current_x = initial_x

print(f"Starting Gradient Descent from x = {initial_x}")

for i in range(num_iterations):
    # calculate the gradient descent at the current x

    grad_at_current_x = gradient(current_x)

    current_x = current_x - learning_rate*grad_at_current_x
    cost_function_at_x = cost_function(current_x)

    # store history
    x_history.append(current_x)
    cost_history.append(cost_function_at_x)


    if (1+1) % 10 == 0 or i == 0:
        print(f"Iteration {i+1}: x= {current_x:.4f}, Cost = {cost_function_at_x:.4f}")
print(f"\nOptimization Finished. final x = {current_x:.4f}")
print(f"\nMinimum cost Found. final x = {cost_function_at_x:.4f}")


x_values = np.linspace(-12,12, 400)
y_values = cost_function(x_values)

plt.figure(figsize=(10,6))
plt.plot(x_values, y_values, label= 'J(X) = x^2', color='blue')

plt.scatter(x_history, cost_history, color='red', s=50, zorder=5, label="GD steps")

plt.plot(x_history, cost_history, color='orange', linestyle='--', linewidth=3, label="GD Path")

plt.title('Gradient Descent for J(X) = X^2')

plt.xlabel('x (Parameter)')
plt.ylabel('J(x) (Cost)')

plt.grid(True)
plt.legend()
plt.annotate(
    'Start', xy=(x_history[0], cost_history[0]), xytext=(x_history[0] + 2, cost_history[0] + 50),
    arrowprops=dict(facecolor='green', shrink=0.05),
    fontsize=10

)

plt.annotate(
    'End', xy=(x_history[-1], cost_history[-1]), xytext=(x_history[-1] - 3, cost_history[-1] + 50),
    arrowprops=dict(facecolor='green', shrink=0.05),
    fontsize=10
)
plt.show()

```
