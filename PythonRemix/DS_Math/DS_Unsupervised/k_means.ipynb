# K-Means

### 1. The Theory of K-Means Clustering 🧩

**Clustering** is a fundamental **unsupervised learning algorithm**. Unlike the previous models we've discussed, it does not rely on labeled data. Its goal is to group similar data points together into clusters, where data points within a cluster are more similar to each other than to data points in other clusters.

**K-Means** is the most common and simple clustering algorithm. The "K" in K-Means represents the number of clusters you want the algorithm to find.

#### How It Works: The Iterative Process

The algorithm works in an iterative, two-step process to find the best cluster centroids.

1.  **Initialization:** The algorithm starts by randomly selecting **K** data points from the dataset to serve as the initial **centroids**. 

2.  **Assignment Step:** For each data point in the dataset, the algorithm calculates its distance to each of the K centroids. It then **assigns** the data point to the cluster of the closest centroid.

3.  **Update Step:** After all data points have been assigned, the algorithm recalculates the position of each of the K centroids. The new centroid position is the **mean** (the average) of all the data points currently assigned to that cluster.

4.  **Convergence:** The algorithm repeats steps 2 and 3 until the centroids no longer move significantly or a maximum number of iterations is reached.

#### The Goal: Minimizing Inertia

The ultimate objective of the K-Means algorithm is to minimize a metric called **inertia**. Inertia measures the sum of squared distances of samples to their closest cluster center. In simpler terms, it's the total "tightness" of the clusters. A lower inertia value means the clusters are more dense and well-separated.

$Inertia = \sum_{i=1}^{m} \min_{j=1}^{K} (||x_i - \mu_j||^2)$

* $x_i$: The data point.
* $\mu_j$: The centroid of cluster $j$.
* $K$: The number of clusters.

This concludes the theory of K-Means. The key takeaways for an interview are that it's an **unsupervised** algorithm that works by iteratively assigning data points to the closest centroids and then moving the centroids to the mean of their assigned points.

Ready to move on to the next part, implementing a simplified K-Means from scratch?

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

# 1. Generate a synthetic dataset for clustering
# make_blobs creates a dataset with distinct, spherical clusters.
X, y = make_blobs(n_samples=500, centers=4, random_state=42)

# 2. Instantiate the K-Means model
# We must specify the number of clusters (n_clusters).
# n_init='auto' handles a common warning.
model = KMeans(n_clusters=4, n_init='auto', random_state=42)

# 3. Fit the model to the training data
# Note that we only pass the features (X), not the labels (y),
# because K-Means is an unsupervised algorithm.
print("Fitting the K-Means model...")
model.fit(X)
print("Fitting complete.")

# 4. Get the cluster assignments and the final centroid positions
# The `labels_` attribute stores the cluster to which each data point was assigned.
cluster_labels = model.labels_
# The `cluster_centers_` attribute stores the final positions of the centroids.
centroids = model.cluster_centers_

# 5. Print the final inertia
# Inertia measures how "tight" the clusters are. Lower is better.
print(f"\nFinal Model Inertia: {model.inertia_:.2f}")

# 6. Visualize the clusters and centroids
plt.figure(figsize=(10, 8))

# Plot the data points, colored by their assigned cluster
plt.scatter(X[:, 0], X[:, 1], c=cluster_labels, s=50, cmap='viridis', label='Data Points')

# Plot the final centroids
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', s=200, marker='X', label='Centroids')

plt.title("K-Means Clustering Results")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(True)
plt.show()

```
