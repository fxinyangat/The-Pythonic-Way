#### CSV

```python
import pandas as pd
import os

# --- Create a dummy CSV file for demonstration ---
csv_data = """Name,Age,City,IsStudent,Score,DateJoined
Alice,24,New York,True,85.5,2023-01-15
Bob,30,Los Angeles,False,92.0,2022-11-20
Charlie,22,Chicago,True,78.2,2024-03-01
David,,"Houston",False,60.0,2023-09-10
Eve,28,Seattle,True,95.0,2024-06-05
Frank,35,Boston,False,NaN,2021-07-22
"""
with open("sample_data.csv", "w") as f:
    f.write(csv_data)

# --- Basic Reading ---
print("--- Basic CSV Read ---")
df_basic = pd.read_csv("sample_data.csv")
print(df_basic)
# Observations:
# - Pandas infers data types (e.g., Age as int, Score as float).
# - It handles missing values (David's age, Frank's score) as NaN.
# - DateJoined is read as an object (string) by default.

# --- Reading with Specific Arguments (for robustness and efficiency) ---
print("\n--- Advanced CSV Read ---")
df_advanced = pd.read_csv(
    "sample_data.csv",
    sep=',',               # Explicitly set the separator (default is comma)
    header=0,              # Row 0 contains column names (default)
    names=['PersonName', 'YearsOld', 'Location', 'CurrentlyStudent', 'ExamScore', 'EnrollmentDate'], # Custom column names
    index_col=False,       # Don't use any column as index (default)
    dtype={                # Force specific data types
        'YearsOld': 'Int64', # Use nullable integer type to allow NaN
        'CurrentlyStudent': bool,
        'ExamScore': float
    },
    na_values=['N/A', 'Unknown', ''], # Treat these strings as NaN
    parse_dates=['EnrollmentDate'], # Parse this column as datetime objects
    comment='#'            # Ignore lines starting with '#'
)
print(df_advanced)
print("\nData types after advanced read:")
print(df_advanced.dtypes)
# Observations:
# - Column names are custom.
# - Age is now 'Int64' (Pandas nullable integer) to accommodate NaN.
# - IsStudent is bool.
# - DateJoined is now datetime64[ns]
# - Empty string for David's age is correctly parsed as NaN (by 'Int64').

# Clean up dummy file
os.remove("sample_data.csv")
```

### JSON (JavaScript Object Notation)

A. Loading JSON with json.load() and json.loads()
json.load(file_object): Reads a JSON object from a file-like object.

json.loads(string): Reads a JSON object from a string.

```python
# resading json data

import json
import os


# dummy 
json_data_str = """
{
    "students": [
        {"id": "s001", "name": "Alice", "grades": {"math": 90, "science": 88}},
        {"id": "s002", "name": "Bob", "grades": {"math": 75, "science": 80}},
        {"id": "s003", "name": "Charlie", "grades": {"math": 95, "science": 92}}
    ],
    "course": "Data Science Fundamentals",
    "year": 2024
}
"""

with open("students_data.json", "w") as f:
    f.write(json_data_str)

print("____Loading JSOn from file___")

with open("students_data.json", 'r') as f:
    data_from_file = json.load(f)

print(f"Loaded data type: {type(data_from_file)}")
print(f"Course: {data_from_file['course']}")

print(f"First student's name: {data_from_file['students'][0]['name']}")

# --- Loading from a string ---
print("\n--- Loading JSON from String ---")
json_string = '{"product": "Laptop", "price": 1200.50, "in_stock": true}'
data_from_string = json.loads(json_string)
print(f"Product from string: {data_from_string['product']}")

# Clean up dummy file
os.remove("students_data.json")

```

#### Dumping JSON with json.dump() and json.dumps()
json.dump(obj, file_object): Writes a Python object to a file-like object as JSON.

json.dumps(obj): Converts a Python object to a JSON formatted string.

```python
import json
import os

config_data = {
    "database": {
        "host": "localhost",
        "port": 5432,
        "user": "admin",
        "password": "securepassword"
    },
    "api_keys": [
        {"service": "weather", "key": "abc123xyz"},
        {"service": "maps", "key": "def456uvw"}
    ],
    "debug_mode": True,
    "version": 1.0
}
# dumping to json file

with open("app_config.json", "w") as f:
    json.dump(config_data, f, indent=4)

print("data written to 'app_config.json")

# --- Dumping to a string ---
print("\n--- Dumping JSON to String ---")
json_output_string = json.dumps(config_data, indent=2)
print(json_output_string)

# Clean up dummy file
os.remove("app_config.json")
```
