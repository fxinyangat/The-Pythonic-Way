####  Iterators and Generators
An iterator is asn object that represents a stream of data.

The Iterator Protocol:
In python an object is an iterator if it implements two special mothods.
1. __iter__(self): this returns the iterator object its self
2. __next__(self) -- this returns the next item in the iteration


##### Understanding Generators with Yield Keyword                   
Generators are a simpler, cleaner way to create iterators. They are functions that contain one or more yield statements.

- The yield statement pauses the function's execution returns the value to the caller an d remembers its state.
- When next is called again, the function resumes where it left of after the yield statement, with local and envt variables intact.

Yield vs Return
- Return terminates a function and sends back a single value
While Yield pauses a function, returns a value and allows it to resume later from the same point. A generator function can yield multiple times.





```python
# Simple count down generator function

def count_down(num):
    while num > 0:
        yield num
        num -= 1
    print("Count Down complete")
counter_gen = count_down(10)



for i in count_down(10):
    print(i)
```

```python
# generate squares upto 100

def generate_squares(num):
    for i in range(num):
        yield i * i

# using generator

for square in generate_squares(200):
    print(square)
    if square >= 100:
        break
```

```python
# simulate a very large file write

def write_large_file(file_name, num_lines):
    with open(file_name, 'w') as f:
        for i in range(num_lines):
            f.write(f"Line {i}: This is a very large fine. \n")
write_large_file('huge_log.txt', 10*7) #create 70 milion lines in a file
```

```python
def read_large_file(file_name):
    with open(file_name, 'r') as f:
        for line in f:
            yield line.strip()

line_count = 0

for line in read_large_file("huge_log.txt"):
    line_count += 1
    print(f"Processing line {line}")
    print(f"PRocessed, {line_count} lines")
```

```python
def count_up_to(max):
    n = 1
    while n <= max:
        yield n
        n += 1

gen = count_up_to(3)
print(next(gen))  # 1
print(next(gen))  # 2
print(next(gen))  # 3
```

```python
def fibonacci_sequence(n):
    a, b = 0, 1
    count = 0
    while count < n:
        yield a
        a, b = b, a + b
        count += 1

print("Fibonacci Sequence (first 10):")
for num in fibonacci_sequence(10):
    print(num)

print("\nFibonacci Sequence (first 5):")
fib_gen = fibonacci_sequence(5)
print(next(fib_gen))
print(next(fib_gen))
print(next(fib_gen))
print(next(fib_gen))
print(next(fib_gen))
# print(next(fib_gen)) # This would raise StopIteration
```

```python
# even numbers in a range generator

def even_numbers(start, end):
    current = start if start % 2 == 0 else start + 1

    while current < end:
        yield current
        current += 2

print("Even Nummbers from 1 to 10")

for num in even_numbers(90,120):
    print(num)
```

```python
# Reverse String Generator

def reverse_str(string_word):
    index = len(string_word) - 1

    while index >= 0:
        yield string_word[index]

        index -= 1
for char in reverse_str("Pythin is Awesome"):
    print(char)


```
