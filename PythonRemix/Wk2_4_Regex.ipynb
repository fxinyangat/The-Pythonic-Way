### Regular Expressions
Regular expressions is a sequence of characters that form a search pattern. Regex can be 
used to check if a string contains a specified pattern.

Why use Regex:
- Find specific patterns
- Validate formats
- Extract Structured data.
- Transfor  text by replacing parts of a string that match the pattern


Common Regex Characters and Constructs:

Literal Characters: Match themselves (e.g., abc matches "abc").

. (Dot): Matches any single character (except newline, by default).

* (Asterisk): Matches the preceding character zero or more times. (e.g., a*b matches "b", "ab", "aab", "aaab").

+ (Plus): Matches the preceding character one or more times. (e.g., a+b matches "ab", "aab", "aaab" but not "b").

? (Question Mark): Matches the preceding character zero or one time. (e.g., ab?c matches "ac", "abc").

{n} (Exactly n times): Matches the preceding character exactly n times. (e.g., a{3} matches "aaa").

{n,} (n or more times): Matches the preceding character n or more times. (e.g., a{2,} matches "aa", "aaa", "aaaa").

{n,m} (n to m times): Matches the preceding character at least n times and at most m times. (e.g., a{2,4} matches "aa", "aaa", "aaaa").

[] (Character Set): Matches any one of the characters inside the brackets. (e.g., [aeiou] matches any vowel).

[a-z]: Any lowercase letter.

[A-Z]: Any uppercase letter.

[0-9]: Any digit.

[a-zA-Z0-9]: Any alphanumeric character.

[^abc]: Matches any character except 'a', 'b', or 'c'.

| (Alternation/OR): Matches either the expression before or after the |. (e.g., cat|dog matches "cat" or "dog").

() (Grouping): Groups expressions together. Also used for capturing (more on this below).

\ (Escape Character): Used to escape special characters (e.g., \. matches a literal dot, \$ matches a literal dollar sign). Also used for special sequences:

\d: Matches any digit ([0-9]). \D matches any non-digit.

\w: Matches any "word" character (alphanumeric + underscore: [a-zA-Z0-9_]). \W matches any non-word character.

\s: Matches any whitespace character (space, tab, newline, etc.). \S matches any non-whitespace.

\b: Matches a word boundary (the position between a word character and a non-word character).

^ (Caret): Matches the start of the string (or start of a line in multiline mode).

$ (Dollar): Matches the end of the string (or end of a line in multiline mode).

Key Functions in Python's re Module:

re.search(pattern, string): Scans through the entire string looking for the first location where the regex pattern produces a match. Returns a match object if found, None otherwise.

re.match(pattern, string): Similar to search, but only checks for a match at the beginning of the string. Returns a match object or None. (Less commonly used than search for general pattern finding).

re.findall(pattern, string): Returns a list of all non-overlapping matches in the string. If the pattern has capturing groups, it returns a list of tuples (one tuple per match, with each tuple containing the captured groups).

re.sub(pattern, repl, string, count=0): Replaces all (or count number of) occurrences of the pattern in string with repl (replacement string).

```python
import re

text = "My phone number is 123-456-7890. Call me at 987.654.3210 or email me at user@example.com. Another number: (555) 111-2222."


# check for a specific word (case-insensitive)

if re.search(r"phone", text, re.IGNORECASE):
    print("Found phone in text")

#  find all simple characters (one or more digits)

numbers = re.findall(f"\d+", text)
print(f"All simple numbers: {numbers}") 

# find words ending 'ing'
words = "Running, jumping and sleeeping are fun"
words_ending_ing = re.findall(r"\w+ing\b", words)
print(f"Words ending with 'ing': {words_ending_ing}")

# 4. Replace multiple spaces with a single space
cleaned_text = re.sub(r"\s+", " ", "This   text      has   too  many   spaces.")
print(f"Cleaned text: '{cleaned_text}'") # Output: 'This text has too many spaces.'

# 5. Validate a simple date format (dd-mm-yy)
date1 = "25-01-2023"
date2 = "5-1-2023"

date_pattern = r"\d{2}-\d{2}-\d{4}"


print(bool(re.match(date_pattern, date1)))

print(bool(re.match(date_pattern, date2)))




```

```python
# Extracting a specific phone number
phone_text = "Contact support at 123-456-7890 for assistance."
# Pattern to capture three groups separated by hyphens
phone_pattern = r"(\d{3})-(\d{3})-(\d{4})"

match = re.search(phone_pattern, phone_text)
if match:
    print(f"\nFull phone number: {match.group(0)}") # The entire match
    print(f"Area code: {match.group(1)}")
    print(f"First 3 digits: {match.group(2)}")
    print(f"Last 4 digits: {match.group(3)}")
    # Output:
    # Full phone number: 123-456-7890
    # Area code: 123
    # First 3 digits: 456
    # Last 4 digits: 7890

# Extracting data from log lines using named groups
log_line = "2025-07-07 15:10:30 [INFO] User 'Inyangat Francis Xavier' logged in from 192.168.1.100."
log_pattern = r"(?P<timestamp>\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2}) \[(?P<level>\w+)\] (?P<message>.+)"

match = re.search(log_pattern, log_line)
if match:
    print(f"\nLog Timestamp: {match.group('timestamp')}") # Access by name
    print(f"Log Level: {match.group('level')}")
    print(f"Log Message: {match.group('message')}")
    # Output:
    # Log Timestamp: 2025-07-07 15:10:30
    # Log Level: INFO
    # Log Message: User 'Inyangat Francis Xavier' logged in from 192.168.1.100.

# Extracting multiple email addresses (re.findall with capturing groups)
email_text = "My email is user1@domain.com and also user2@mail.org. Invalid: invalid@.com"
email_pattern = r"(\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,7}\b)" # Simple email pattern

emails_found = re.findall(email_pattern, email_text)
print(f"\nEmails found: {emails_found}")
# Output: ['user1@domain.com', 'user2@mail.org'] (Note: findall returns strings for simple patterns/single groups)

# If multiple capturing groups are used with findall, it returns a list of tuples:
pattern_multi_group = r"(\d{4})-(\d{2})-(\d{2})"
dates_in_text = "Dates: 2023-01-15, 2024-11-30."
all_dates = re.findall(pattern_multi_group, dates_in_text)
print(f"All dates (year, month, day tuples): {all_dates}")
# Output: [('2023', '01', '15'), ('2024', '11', '30')]
```

```python
# Practice

# Phone number Normalization

phone_numbers = ["(123) 456-7890", "987.654.3210", "555-111-2222", "800-CALL-NOW"]

phone_pattern = r"\D*(\d{3})\D*(\d{3})\D*(\d{4})"

normalized_phone_nums = []

for num in phone_numbers:
    normalized = re.sub(phone_pattern,r"\1-\2-\3", num)
    normalized_phone_nums.append(normalized)
    print(f"Original: '{num}' -> Normalized: '{normalized}'")
```

```python
# Product code extraction

inventory = "Product A: XYZ-1001; Product B: MNO-2002; Product C: PQR-3003."

pattern_prod = r"[A-Z]{3}-\d{4}"

products_codes = re.findall(pattern_prod, inventory)
print(products_codes)
```

```python
# Basic URL Validation

# For each url, determine if its https or http then extraxt the domain name.

urls = [
    "http://example.com",
    "https://www.google.com/search?q=regex",
    "ftp://archive.org",
    "invalid-url",
    "https://developer.mozilla.org/en-US/docs/Web/HTTP/Headers",
    "HTTP://anothersite.net" # Test case for case insensitivity
]
url_pattern = re.compile(r"^(?P<protocol>https?)://(?P<domain>[a-zA-Z0-9.-]+)(?:/.*)?$", re.IGNORECASE)

extracted_urls = []

for url in urls:
    match = url_pattern.search(url)

    if match:
        extracted_info = (match.group('protocol').lower(), match.group('domain').lower())
        extracted_urls.append(extracted_info)
        print(f"'{url}' -> Protocol: '{extracted_info[0]}', Domain: '{extracted_info[1]}'")
    else:
        extracted_urls.append(None)
        print(f"'{url}' -> No valid HTTP/HTTPS URL found.")

print(f"\nCollected extracted data: {extracted_urls}")
```
