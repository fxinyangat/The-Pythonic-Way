```python
import pandas as pd
import numpy as np

sales_df = pd.DataFrame({
    'OrderID': range(1, 11),
    'CustomerID': [101, 102, 101, 103, 104, 105, 102, 101, 103, 106],
    'Product': ['Laptop', 'Mouse', 'Monitor', 'Keyboard', 'Laptop', 'Mouse', 'Monitor', 'Laptop', 'Keyboard', 'Mouse'],
    'Quantity': [1, 2, 1, 1, 1, 3, 1, 2, 2, 1],
    'Price': [1200, 25, 250, 75, 1200, 25, 250, 1200, 75, 25],
    'ShippingDate': pd.to_datetime(['2025-01-05', '2025-01-06', '2025-01-07', '2025-01-08', '2025-01-09',
                                    '2025-01-10', '2025-01-11', '2025-01-12', '2025-01-13', '2025-01-14']),
    'Discount': [0.1, 0.05, 0.1, 0, 0.1, 0, 0.05, 0.15, 0.1, 0]
})

# Apply previous corrections
sales_df.loc[2, 'Product'] = 'Monitor'
sales_df.loc[6, 'Price'] = 250
sales_df['Product'] = sales_df['Product'].str.lower().str.strip()
sales_df['TotalRevenue'] = sales_df['Quantity'] * sales_df['Price']

# Let's adjust the indices to be cleaner for this round
sales_df = sales_df.set_index('OrderID')

print("--- Cleaned and Prepped DataFrame for this round ---")
print(sales_df)
```

```python
# units of each product ordered by each customer

order_stats = sales_df.groupby(['Product','CustomerID']).agg(
    no_units = ('Quantity','sum'),
    total_spend = ('TotalRevenue', 'sum'),
    orders_placed = ('CustomerID', 'count')


).reset_index().sort_values(by='total_spend', ascending=False)
order_stats
```

```python
# convert stats to pivot

order_stats_pivot = order_stats.pivot_table(
    index="CustomerID",
    columns="Product",
    values="no_units",
    aggfunc="sum",
    fill_value=0
)

order_stats_pivot
```

```python
# using a pivot table. best approach

customer_product_units = sales_df.pivot_table(
    index='CustomerID',
    columns='Product',
    values='Quantity',
    aggfunc='sum',
    fill_value=0
    
)
customer_product_units
```

```python
# products with high discount
disc_check=lambda x: 1 if x>=0.1 else 0
sales_df['HighDiscountFlag'] = sales_df['Discount'].apply(disc_check)
sales_df

# better alternative

sales_df['HighDiscountFlag2'] = np.where(sales_df['Discount']>=0.1, 1,0)
sales_df

sales_df['Price_Quartile'] = pd.qcut(sales_df['Price'], q=4, labels=['Low', 'Medium', 'High', 'Very High'])
sales_df.head()


sales_df['Price_Quartile_Manual'] = pd.cut(sales_df['Price'], bins=[0,400,600,800,1200], labels=['Low', 'Medium', 'High', 'Very High'] ) 
sales_df
```

```python
# average total revene for all customer

# Step 1: Calculate the total revenue for each customer
customer_total_revenue = sales_df.groupby('CustomerID')['TotalRevenue'].sum()

print("--- Total Revenue per Customer ---")
print(customer_total_revenue)
print("\n")

# Step 2: Calculate the average of those total revenues
average_customer_revenue = customer_total_revenue.mean()

print("--- Average Total Revenue per Customer ---")
print(f"${average_customer_revenue:.2f}")
```
