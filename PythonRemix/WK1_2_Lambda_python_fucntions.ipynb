### Lambda Functions, Python built in Functions
- A lambda function is called "anonymous" because it does not have a formal name.
- This makes them ideal for situations where a function is only needed for a very short period, often as an argument to another function, and giving it a formal name would be unnecessary boilerplate.
- The result of this single expression is automatically returned by the lambda function, without needing an explicit return keyword.

syntax:
lambda arguments: expression

When/Why use lambda functions
- Eliminates def boilerplate: For very short functions, lambda saves you the def keyword, function name, colon, and return statement.
- Inline use: They are perfect for situations where a function is passed as an argument to a higher-order function (like map(), filter(), sorted(), min(), max()) and the logic is simple enough to fit on one line.
- Local scope: They are defined and typically used within a small, local scope.

```python
# Example 1 sUM two numbers
def add(a, b):
    return a + b
print(add(2, 3)) # Output: 5

# Lambda version

add_lambda = lambda a,b: a+b
print(add_lambda(4,5))
```

```python
# Example 2: Check Even
def is_even_def(num):
    return num % 2 == 0
print(is_even_def(4)) # Output: True
print(is_even_def(7)) # Output: False

# Lambda version

is_even_lambda = lambda n : n % 2 == 0


is_even_lambda = lambda x: x % 2 == 0
is_even_num = list(map(is_even_lambda, range(8)))

print(is_even_num)
```

```python
# concatenate two strings
check_positive = lambda num: "positive" if num > 0 else "negative"

print(check_positive(-99*-2))
```

### Lambda Functions: Used in map(), filter(), and sorted()
These three built-in Python functions are "higher-order functions" because they can take another function (like a lambda) as an argument. They apply that function to iterables in different ways.

#### Map

Applies a given function to each item of iterable

Syntaxt map(function, iterable)

```python
fahrenheit_temps = [-30, -20, 100, 0, 10, 20, 30]

# --- Using a traditional def function with map() ---
def fahrenheit_to_celsius(f):
    return (5/9) * (f - 32)

# using map

far_to_cel = list(map(lambda f: (5/9) * (f - 32),fahrenheit_temps))

print({f:c for f,c in zip(fahrenheit_temps, far_to_cel)})
```

```python
prices = [100, 50, 45, 8000, 1000]
# apply 10% discount
price_after_discount = [price*0.9 for price in prices]
# add $3.5 sales tax
price_after_tax = list(map(lambda p: p + 3.5, price_after_discount))

print(price_after_discount)

print(price_after_tax)

# before vs after

price_dict = {price_b4: price_after for price_b4, price_after in zip(prices, price_after_tax)}

print(price_dict)
```

```python
# filter words shorter than 5
words = ["apple", "banana", "kiwi", "grapefruit", "cat"]

def word_len(word):
    
    return len(word) > 5
print(list(filter(word_len, words)))

short_words = list(filter(lambda s: len(s) <= 5, words))
short_words
```

```python
# sorted

data = [('apple', 3), ('banana', 1), ('cherry', 2)]

# sort by second element


sorted_data_lambda = sorted(data, key = lambda i: i[1])
sorted_data_lambda

# sord words by their length, longest first

words = ["apple", "kiwi", "grapefruit", "banana"]

sorted_words_lambda = sorted(words, key = lambda word: len(word), reverse=True)
sorted_words_lambda
```

```python
# Apply 5% saless tax
prices = [10.50, 25.00, 3.75, 50.25]

after_tax = list(map(lambda price: price*1.05, prices))
after_tax

# full names
names = [("John", "Doe"), ("Alice", "Smith")]

full_names = list(map(lambda x: x[0]+ " " + x[1], names))
full_names
```
