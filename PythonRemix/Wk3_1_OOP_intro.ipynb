## Classes and Objects

# OOP (Object Oriented Programming)

OOP is a programming paradigm based on the concept of "objects," which can contain data (attributes) and code (methods). The main goal of OOP is to bind together the data and the functions that operate on them so that no other part of the code can access that data except that function.

The core principles of OOP are:

Encapsulation: Bundling data and methods that operate on that data within a single unit.

Inheritance: Allowing new classes to inherit properties and behaviors from existing classes.

Polymorphism: The ability of objects of different classes to respond to the same method call in a way that is specific to their class.

Abstraction: Hiding complex implementation details and showing only the necessary features of an object.


### Classes and Objects

#### Classes
At the heart of OOP are classes and objects. They are fundamental concepts that model real-world (or abstract) entities in your code.

##### 1.1. What are Classes? (The Blueprint)
A class is a blueprint, a template, or a prototype from which objects are created. It defines a set of attributes (data) that the objects will possess and a set of methods (functions) that the objects can perform.

Think of it like this:

The blueprint for a house is a class. It specifies that every house will have walls, a roof, rooms, and methods like "open door" or "turn on lights."

The blueprint itself is not a house you can live in; it's just the design.

##### 1.2. What are Objects? (The Instance)
An object (also known as an instance) is a concrete realization of a class. It's a specific entity created based on the class's blueprint. Each object has its own unique set of attribute values.

Continuing the metaphor:

An actual house built according to that blueprint is an object. You can live in that specific house, with its particular paint color, number of windows, and whether its lights are currently on or off. You can have multiple houses (objects) built from the same blueprint (class), but each house is distinct.

```python
class Dog:
# a simple class to represent a dog
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age 

# object instantiation - creating objects of the Dog class.
my_dog1 = Dog("Bunny", 3)

print(f"Object created: {my_dog1}")
print(f"Dog 1 name is {my_dog1.name}")
print(f"Dog 1 age is {my_dog1.age}")

print("__"*20)

my_dog2 = Dog("Boogy", 5)


print(f"Object created: {my_dog2}")
print(f"Dog 2 name is {my_dog2.name}")
print(f"Dog 2 age is {my_dog2.age}")

# We can see they are distinct objects with their own data
print(f"Are my_dog1 and my_dog2 the same object? {my_dog1 is my_dog2}")
print(f"my_dog1's name: {my_dog1.name}, my_dog2's name: {my_dog2.name}")

# Even if they had the same name/age, they would be distinct objects
my_dog3 = Dog("Bunny", 3)
print(f"my_dog1 == my_dog3? {my_dog1 is my_dog3}") # Still False, they are different objects in memory

        
```

### Concept 2: Attributes and Methods

Every object is essentially a bundle of data and the functions that operate on that data.

Attributes: These are the data or properties associated with an object. They describe the object's state.

Methods: These are the functions or actions that an object can perform. They define the object's behavior.


**2.1. Instance Variables (Attributes)**
We touched on these briefly with the __init__ method. Instance variables are attributes whose values are unique to each instance (object) of a class. They are defined inside the __init__ method (or sometimes added later outside __init__ if needed, though less common for initial setup) and are always accessed using the self keyword.

Example from Dog: self.name and self.age were instance variables because each Dog object has its own specific name and age.

**2.2. Class Variables (Attributes)**
Class variables are attributes that are shared by all instances of a class. They belong to the class itself, not to any particular object. You define them directly inside the class body, but outside any method.

Use Case: Ideal for data that is constant for all objects of that class, or for tracking general information about the class (e.g., a count of how many objects have been created).

Access: You can access them using either the class name (e.g., Dog.species) or an instance name (e.g., my_dog.species), but it's generally best practice to use the class name for clarity when referring to the class-level attribute.

**2.3. Instance Methods**
Instance methods are functions defined inside a class that operate on the data (instance variables) of a specific object. They always take self as their first parameter, which refers to the instance on which the method is called.

```python
# Example

print("______Concept 2: Attributes and Methods___________")

class Dog:
    # class variable shared by all instances of the Dog Class
    species = "Canis familiaris"
    def __init__(self, name: str, age:int, breed: str):
        self.name = name
        self.age = age
        self.breed = breed
        self.is_happy = True #Default

    def bark(self):
        if self.age < 2:
           return f"{self.name} says 'yap yap'"
        else:
            return f"{self.name} says 'wooof woof'" 
    def get_info(self):
        return f"Dog name is {self.name}. Its a {self.age} year old {self.breed} breed. Is Happy: {self.is_happy is True} "
    
    def make_unhappy(self):
        self.is_happy = False
        return f"{self.name} is now UN Happy"
       
    
# --- Object Instantiation and Usage ---

# Create Dog objects
dog1 = Dog("Buddy", 3, "Golden Retriever")
dog2 = Dog("Lucy", 5, "Labrador")
dog3 = Dog("Max", 1, "Pug")


print(f"Dog 1 info: {dog1.get_info()}")
print(f"Dog 2 info: {dog2.get_info()}")
print(f"Dog 3 info: {dog3.get_info()}")


print("-" * 20)

# Accessing Class Variable
print(f"Dog 1's species (via instance): {dog1.species}")
print(f"Dog 2's species (via instance): {dog2.species}")
print(f"Class Dog's species (via class): {Dog.species}") # Preferred way to access class variables

print("-" * 20)

# Calling Instance Methods
print(dog1.bark()) # Buddy (3 years old) says Woof! Woof!
print(dog2.bark()) # Lucy (5 years old) says Woof! Woof!
print(dog3.bark()) # Max (1 year old) says Yap! Yap!

print("-" * 20)

# Demonstrating unique instance variables
print(f"Is {dog1.name} happy? {dog1.is_happy}")
dog1.make_unhappy()
print(f"Is {dog1.name} happy now? {dog1.is_happy}")
print(f"Is {dog2.name} happy? {dog2.is_happy}") # Lucy is still happy, as is_happy is an instance variable


```

**Key Takeaways from this concept:**

Instance Variables (self.attribute): Store data unique to each object. Defined and initialized typically in __init__.

Class Variables (ClassName.attribute): Store data shared by all objects of the class. Defined directly within the class body.

Instance Methods (def method(self, ...)): Define behaviors for individual objects. They require self as the first parameter to access the object's unique data.

**Changing Class and instance variables**

1. Direct Modification (Recommended): You can change a class variable by assigning a new value to it using the class name: ClassName.class_variable = new_value. This is the standard way to modify it, and the change will apply to all existing and future instances that don't have their own instance variable shadowing it.

2. Instance-level Shadowing (Caution): If you assign a value to a class variable using an instance (instance_object.class_variable = new_value), you don't actually change the class variable. Instead, you create a new instance variable with the same name, which "shadows" the class variable for that specific instance. Other instances (and the class itself) will still see the original class variable.

```python
class Animal :
    species = "mamals"
    def __init__(self, name:str, breed: str):
        self.name = name
        self.breed = breed

    def get_animal_details(self):
        return f"{self.name} belongs to {self.breed} breed and is a species of {Animal.species}"
    

# object instantiation

animal1 = Animal("goat", "kikuyu")
animal2 = Animal("lion", "African Lion")

print(animal1.get_animal_details())
print(animal2.get_animal_details())

print(f"Animal is of Species {Animal.species}")

Animal.species = "birds"
animal1.species = "mamal"
print(f"Animal Species changed to {Animal.species}")
print(f"Animal Species changed to instance level {animal2.species}")

del animal1.species
print(f"dog1's species after deleting its instance variable: {animal1.species} (back to class variable)") # Output: Canis lupus familiaris

```

### Inheritance



**Concept 3: Inheritance**
Inheritance is a mechanism in OOP that allows a new class to inherit properties (attributes) and behaviors (methods) from an existing class. This creates a hierarchical relationship between classes

**Benefits of Inheritance**
Code Reusability: Avoids duplicating code. Common functionality is defined once in the parent class and reused in multiple child classes.

Modularity: Organizes code into logical, hierarchical structures.

Extensibility: Easier to extend existing functionality by creating new subclasses without modifying the parent class.



```python
# super class
class Animal:
    def __init__(self, name: str, species:str):
        self.name = name
        self.species = species
        self.is_alive = True

    def make_sound(self):
        return f"This is a generic sound of all animals."
    
    def get_animal_info(self):
        return f"name: {self.name}, species: {self.species}"

# child class (sub-class)

class Dog(Animal):
    def __init__(self, name:str, age:int, breed: str ):
        super().__init__(name, "Dog")

        self.age = age
        self.breed = breed
        self.loyalty = 10

    # method overiding
    def make_sound(self):
        return f"{self.name} says woof woof"

    def fetch(self, item: str):
        return f"{self.name} fetched {item}"
    
    def get_animal_info(self):
        base_info = super().get_animal_info()
        return f"{base_info}. they are {self.age} year old. Breed {self.breed} and Loyalty Rate {self.loyalty}"

# sub class

class Cat(Animal):
    def __init__(self, name:str, color: str):
        super().__init__(name, "Cat")
        self.color = color
        self.has_attitude = True
    
    def make_sound(self):
        return f"{self.name} has said meeeww meew"

    def scratch(self, item:str):
        return f"{self.name} has scratch the {item}"


print("\n--- Creating Objects ---")
my_dog = Dog("Buddy", 4, "Golden Retriever")
my_cat = Cat("Whiskers", "Tabby")
generic_animal = Animal("Leo", "Lion")


print("\n--- Testing inherited and overridden methods ---")
print(my_dog.get_animal_info())    # Calls Dog's overridden get_animal_info
print(my_dog.make_sound())  # Calls Dog's overridden make_sound
print(my_dog.fetch("ball")) # Calls Dog's specific method

print("-" * 10)

print(my_cat.get_animal_info())    # Calls Cat's inherited get_animal_info (from Animal)
print(my_cat.make_sound())  # Calls Cat's overridden make_sound
print(my_cat.scratch("sofa")) # Calls Cat's specific method

print("-" * 10)

print(generic_animal.get_animal_info())    # Calls Animal's get_animal_info
print(generic_animal.make_sound())  # Calls Animal's make_sound

print("\n--- Demonstrating inherited attributes ---")
print(f"{my_dog.name} is alive: {my_dog.is_alive}")      # Inherited from Animal
print(f"{my_cat.name} is alive: {my_cat.is_alive}")      # Inherited from Animal
print(f"{generic_animal.name} is alive: {generic_animal.is_alive}") #


        


    




```

## Polymorphism

Concept 4: Polymorphism
Polymorphism literally means "many forms." In OOP, it refers to the ability of different objects to respond to the same message (method call) in their own unique way. This means you can write code that operates on objects of various classes as long as those classes share a common interface (i.e., they have methods with the same names).

**4.1. The Essence of Polymorphism**
Imagine you have a button labeled "Play Sound."

If you press it on a Dog object, it barks.

If you press it on a Cat object, it meows.

If you press it on a Bird object, it chirps.

The action (pressing the "Play Sound" button/calling make_sound()) is the same, but the result varies depending on the type of object you're interacting with.

**4.2. Polymorphism and Inheritance**
Polymorphism is often achieved through inheritance. When subclasses override methods from their parent class, they provide their own "form" of that method's behavior. This allows you to treat objects of different subclasses uniformly through a common superclass interface.



```python
class Animal:
    def __init__(self, name:str, species:str):
        self.name = name
        self.species = species
    
    def make_sound(self):
        return "Generic Animal Sound"
    
    def get_info(self):
        return f"{self.name} is a {self.species}"
    
class Dog(Animal):
    def __init__(self, name: str, age:int, breed:str):
        super().__init__(name, "Dog")
        self.age = age
        self.breed = breed
    
    def make_sound(self): #overrides animal make_sound
        return f"{self.name} barks: woof! woof!"
    
class Cat(Animal):
    def __init__(self, name:str, color:str):
        super().__init__(name, "Cat")
        self.color = color

    def make_sound(self):
        return f"{self.name} meows: Meow!"

# Introducing new class Bird, also inherting from Animal

class Bird(Animal):
    def __init__(self, name:str, feather_color:str):
        super().__init__(name, "Bird")
        self.feather_color = feather_color

    def make_sound(self):
        return f"{self.name} chirps: Tweet!, Tweet!"
    
    def fly(self):
        return f"{self.name} flies gracefully with its {self.feather_color} feathers."

# Demonstrating Polymorphism

# Create a list of various animals

animals = [
    Dog("Buddy", 4, "Golden Retriever"),
    Cat("Whiskers", "Tabby"),
    Bird("Chirpy", "Blue"),
    Animal("Leo", "Lion")
]


for animal in animals:
    print(f"{animal.get_info()} and it says: {animal.make_sound()}")

print("-"*30)


# 2: Polymorphic behaviour.

def animal_action(animal_obj):
    print(f"I hear an animal! It's {animal_obj.get_info()}")
    print(f"It says: {animal_obj.make_sound()}")

print("-"*10)

print("\n--- Using a Polymorphic Function ---")
animal_action(Dog("Rex", 2, "German Shepherd"))
animal_action(Cat("Mittens", "Black"))
animal_action(Bird("Robin", "Brown"))
animal_action(Animal("Elephant", "Elephant")) # Works for the base class too!

print("-"*10)

# Imagin a non-animal object that can also make sound
class AlarmClock:
    def __init__(self, time_set:str):
        self.time_set = time_set
    
    def make_sound(self):
        return "Riiiing! Riiing! (Time to wake up)"
    
    def get_info(self):
        return f"the alarm clock set for {self.time_set}"
    
my_alarm = AlarmClock("7:00 AM")

animal_action(my_alarm)
    


```


### Concept 5: Encapsulation and Data Hiding
Encapsulation is one of the fundamental principles of OOP. It's the practice of bundling data (attributes) and the methods (functions) that operate on that data into a single unit—the object. It also involves restricting direct access to some of the object's components from outside the object.

Data Hiding is a key aspect of encapsulation. It means making the internal state of an object inaccessible from the outside world directly, forcing external code to interact with the object's data only through its well-defined public methods.

5.1. The Purpose
Data Integrity: Prevents accidental or unauthorized modification of an object's internal data.

Maintainability: Changes to internal implementation details don't affect external code, as long as the public interface remains consistent.

Modularity: Objects become self-contained units that are easier to understand, test, and reuse.

Reduced Complexity: Hides the internal complexity from the user of the object.

5.2. Python's Approach to "Private" Members
Unlike some other languages (like Java or C++) that have explicit private keywords, Python takes a more conventional and less restrictive approach. It relies on naming conventions to suggest how attributes and methods should be accessed. Python has a philosophy of "we're all consenting adults here," meaning if you really want to access a "private" member, Python won't strictly prevent you, but it warns you not to.

A. "Protected" Members (Single Underscore _)
Convention: Attributes or methods prefixed with a single underscore (e.g., _my_attribute, _my_method).

Meaning: This is a convention indicating that the attribute or method is intended for internal use within the class or by its subclasses. It's a hint to other developers: "Don't touch this directly unless you know what you're doing."

Access: It can still be accessed directly from outside the class. Python does not enforce any restriction.

B. "Private" Members (Double Underscore __)
Convention: Attributes or methods prefixed with a double underscore (e.g., __my_attribute, __my_method).

Meaning: This triggers a mechanism called name mangling. Python internally renames these attributes to make them harder (but not impossible) to access directly from outside the class. The primary purpose of name mangling is to prevent name clashes in subclasses, not strict security.

Name Mangling: An attribute like __my_attribute in MyClass becomes _MyClass__my_attribute internally. If you try to access my_object.__my_attribute, you'll get an AttributeError. You can still access it via my_object._MyClass__my_attribute, but this is highly discouraged.

5.3. Properties (@property Decorator)
The most Pythonic and recommended way to achieve controlled access (encapsulation) to attributes is using properties. Properties allow you to define methods (getter, setter, deleter) that are accessed like regular attributes. This means you can add logic (e.g., validation, computation, side effects) when an attribute is read, written, or deleted, without changing the way external code interacts with it.

@property (Getter): Marks a method as a getter for an attribute. When you access obj.attribute_name, this method is called.

@attribute_name.setter (Setter): Defines a method that is called when you assign a value to obj.attribute_name. This is where you can add validation logic.

@attribute_name.deleter (Deleter): Defines a method called when you use del obj.attribute_name.

```python
class Animal:
    def __init__(self, name:str, species:str):
        self.name = name
        self.species = species
    
    def make_sound(self):
        print(f"Animal makes generic Animal sound")

    def move(self):
        print(f"Animal moves like all animals")
    
    def animal_info(self):
        print(f"Name: {self.name} - Species: {self.species}")


class Dog(Animal):
    def __init__(self, name: str, age:int, breed:str):
        super().__init__(name, "Dog")
        self.age = age
        self. breed = breed
    
    

    def make_sound(self):
        print(f"{self.name} is a {self.breed} breed barks: Wooof! Woof!")

    def move(self):
        return print(f"{self.name} walks on four legs")
    

class Bird(Animal):
    def __init__(self, name:str, feather_color:str):
        super().__init__(name, "Bird")
        self.feather_color = feather_color

    def make_sound(self):
        return print(f"{self.name} is a {self.feather_color} colored feathered bird chirps: Tweet! Tweet!")

    def move(self):
        return print(f"{self.name} flys on two wing")
    

animals = [
    Animal("GeneralAnimal","JustAnimal"),
    Bird("Penguin Bird", "Gray White"),
    Dog("Homeguard", 9, "Chuwawa Dog")
]

for animal in animals:
    print("_" * 20)
    print(animal.animal_info())

    print("_" * 20 + " animal sounds")

    animal.make_sound()

    print("_" * 20 + "animal movements")

    animal.move()

def animal_lookalike(animal_object):
    
    print(f"{animal_object.animal_info()}  :animal details")

    
    print(f"{animal_object.make_sound()} : animal sound")
    

# text abstraction

print("testing polymorphism "+  "__"*10)

animal_lookalike(Dog("Pombe", 11, "Dogbreed"))
animal_lookalike(Bird("Pengui Bird", "Gray White"))
animal_lookalike(Animal("GeneralAnimal","JustAnimal"))


class AlarmClock2:
    def __init__(self, set_time: str):
        self.set_time = set_time

    def make_sound(self):
        return print("Riiiiing! Riiiiiing! (Time to wake up)")
    
    def animal_info(self):
        # This method provides the identifying information for the AlarmClock object
        return print(f"The alarm clock set for {self.set_time}")
    
my_alarm2 = AlarmClock2("5:00PM")

animal_lookalike(my_alarm2)

    




    
```

## 5. Encapsulation
Encapsulation is one of the fundamental principles of OOP. It's the practice of bundling data (attributes) and the methods (functions) that operate on that data into a single unit—the object. It also involves restricting direct access to some of the object's components from outside the object.

Data Hiding is a key aspect of encapsulation. It means making the internal state of an object inaccessible from the outside world directly, forcing external code to interact with the object's data only through its well-defined public methods.

**5.1. The Purpose**
Data Integrity: Prevents accidental or unauthorized modification of an object's internal data.

Maintainability: Changes to internal implementation details don't affect external code, as long as the public interface remains consistent.

Modularity: Objects become self-contained units that are easier to understand, test, and reuse.

Reduced Complexity: Hides the internal complexity from the user of the object.

```python
print("--- Concept 5: Encapsulation and Data Hiding ---")

class BankAccount:
    """
    A class representing a bank account, demonstrating encapsulation
    and controlled access to data using properties.
    """
    def __init__(self, account_holder: str, initial_balance: float):
        self.account_holder = account_holder
        # 1. Protected Member: _balance (conventionally for internal use)
        # We store the actual balance in a protected attribute.
        # This is where the real data resides.
        self._balance = initial_balance
        
        # 2. Private Member (Name Mangled): __account_number
        # This is meant to be highly internal and avoid naming collisions in subclasses.
        # It's generated here for demonstration.
        self.__account_number = self._generate_account_number()

    def _generate_account_number(self):
        """
        Protected method: Conventionally for internal use.
        Simulates generating a unique account number.
        """
        import random
        return f"ACC-{random.randint(100000, 999999)}"

    # --- Properties: Controlled access to _balance ---
    # 3. @property (Getter): Allows accessing balance like an attribute (account.balance)
    @property
    def balance(self) -> float:
        """
        Getter for the account balance.
        When you do 'account.balance', this method is called.
        """
        print(f"Accessing balance for {self.account_holder}...")
        return self._balance

    # 4. @balance.setter (Setter): Allows setting balance like an attribute (account.balance = value)
    @balance.setter
    def balance(self, amount: float):
        """
        Setter for the account balance with validation.
        When you do 'account.balance = value', this method is called.
        """
        if not isinstance(amount, (int, float)):
            raise TypeError("Balance must be a number.")
        if amount < 0:
            raise ValueError("Balance cannot be negative.")
        print(f"Setting balance for {self.account_holder} to {amount}...")
        self._balance = amount

    # 5. @balance.deleter (Deleter): Allows deleting balance (del account.balance)
    @balance.deleter
    def balance(self):
        """
        Deleter for the account balance.
        Discourages deleting balance but shows capability.
        """
        print(f"Warning: Deleting balance for {self.account_holder}. Setting to 0.")
        self._balance = 0.0 # Or raise an error

    # Method to interact with balance (public interface)
    def deposit(self, amount: float):
        """Deposits funds into the account."""
        if amount <= 0:
            raise ValueError("Deposit amount must be positive.")
        self.balance += amount # Uses the setter implicitly
        print(f"Deposited {amount}. New balance: {self.balance}") # Uses the getter implicitly

    def withdraw(self, amount: float):
        """Withdraws funds from the account."""
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")
        if self.balance < amount: # Uses the getter implicitly
            raise ValueError("Insufficient funds.")
        self.balance -= amount # Uses the setter implicitly
        print(f"Withdrew {amount}. New balance: {self.balance}") # Uses the getter implicitly

    # Public method to get the account number
    def get_account_number(self):
        """Returns the account number."""
        return self.__account_number # Accessing the name-mangled attribute from inside the class

# --- Usage Demonstration ---

print("\n--- Creating a BankAccount object ---")
my_account = BankAccount("Inyangat Xavier", 1000.0)

print("\n--- Accessing attributes via public interface (properties) ---")
print(f"Account Holder: {my_account.account_holder}")
print(f"Current Balance: ${my_account.balance}") # Calls the @property getter

print("\n--- Modifying balance via setter ---")
my_account.balance = 1200.0 # Calls the @balance.setter
print(f"New Balance after direct set: ${my_account.balance}")

print("\n--- Testing deposit and withdrawal methods ---")
my_account.deposit(200) # Uses the setter and getter
my_account.withdraw(150) # Uses the setter and getter

print("\n--- Testing validation with properties ---")
try:
    my_account.balance = -500 # This will trigger the ValueError in the setter
except ValueError as e:
    print(f"Error setting balance: {e}")

try:
    my_account.withdraw(2000) # This will trigger the ValueError in withdraw method
except ValueError as e:
    print(f"Error withdrawing: {e}")

print("\n--- Demonstrating 'Protected' and 'Private' (Name Mangled) members ---")
# Accessing protected member (by convention discouraged, but possible)
print(f"Direct access to _balance (discouraged): ${my_account._balance}")

# Attempting to access __account_number directly (will cause AttributeError)
try:
    print(f"Direct access to __account_number: {my_account.__account_number}")
except AttributeError as e:
    print(f"AttributeError trying to access __account_number directly: {e}")

# Accessing __account_number via name mangling (possible, but highly discouraged)
# The actual internal name is _ClassName__attribute_name
print(f"Accessing __account_number via mangled name (highly discouraged): {my_account._BankAccount__account_number}")

# Accessing through the public method is the correct way
print(f"Account Number (via public method): {my_account.get_account_number()}")

print("\n--- Deleting the balance property (uses @balance.deleter) ---")
del my_account.balance # Calls the @balance.deleter
print(f"Balance after deletion: ${my_account.balance}")
```

**Key Takeaways from this concept:**

Encapsulation: Bundles _balance, __account_number, deposit, withdraw within BankAccount, making it a self-contained unit.

Data Hiding: By using _balance (protected) and __account_number (name-mangled "private"), we signal that these are internal.

Properties (@property, @setter): This is the Pythonic way to achieve controlled access to attributes.

They allow you to treat methods like attributes, making your interface clean (e.g., account.balance instead of account.get_balance()).

They enable you to embed validation logic (e.g., ensuring balance isn't negative) or computation whenever the attribute is read or written.

This ensures data integrity without exposing the raw internal attribute directly.

Encapsulation, especially through properties, is essential for writing robust and maintainable OOP code in Python. It's about creating clear contracts for how objects can be safely interacted with.
