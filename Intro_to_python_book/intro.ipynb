```python
from functools import reduce
int1 = int(input("Enter first number:"))
int2 = int(input("Enter second number:"))
int3 = int(input("Enter third number:"))

numbers = [int1, int2, int3]

max_num = lambda x,y: x if x > y else y

final_max = reduce(max_num, numbers)
print(numbers)
final_max
```

### while & for loops

```python
numbers = [1,2,3,4,5,66]

length_num = len(numbers)
num_index = 0

while length_num > 0:
    print(numbers[num_index])
    num_index += 1
    length_num -= 1

```

```python
marks = [98, 76, 71, 87, 83, 90, 57, 84, 94]

av_mark = reduce(lambda x,y: x+y, marks)/len(marks)
print(f"{av_mark:.20f}")

import statistics

mode_mark = statistics.mode(marks)
median_mark = statistics.median(marks)
mean_mark = statistics.mean(marks)

print(f"Mode:{mode_mark}\nMean:{mean_mark}\nMedia: {median_mark}")
```

### Game of Dice Rolling

```python
import random

def roll_dice():
    dice1 = random.randrange(1,7)
    dice2 = random.randrange(1,7)
    print(f"Dice 1: {dice1}\nDice2:{dice2}")
    return (dice1,dice2)

def display_dice(dice):
    die1,die2 = dice
    print(f"Palyer rolled {die1} + {die2} = {sum(dice)}")


dice_values = roll_dice() # first roll
display_dice(dice_values)


list1 = [78,23,67,44]

itemindex = list(enumerate(list1))



list1.sort(reverse=1)

list1
```
