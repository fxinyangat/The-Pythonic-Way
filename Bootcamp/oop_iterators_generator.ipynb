```python
class Car:

    def __init__(self, color, model):
        self.color = color #property 1
        self.model = model #property 2
        self.speed = 0 #property 3

    def drive(self): #method 1
        self.speed = 60
        print(f"the {self.color} {self.model} is driving at {self.speed} km/h")
    
    def brake(self): # method 2
        self.speed = 0
        print(f"The {self.color} {self.model} has stopped.")
    

        
```

```python
#create objects from the Car class

car1 = Car("Red", "Toyota")

car2 = Car("BLue", "Honda")

car3 = Car("Black", "Hyundai")


car3.drive() # the Black Hyundai is driving at 60 km/h



car1.brake() #The Red Toyota has stopped.

```

### Multiple inheritance and Multi-level inheritance

In multiple inheritance, one class inherits and another class inherit from the child class.

In multi level inheritance, one class inherits from morethan one class




```python
#Multi level inheritance.

class A:
    def __init__(self,a):
        self.a = a 
    def methA(self):
        print("method of A")
class X:
    def __init__(self, x):
        self.x =x
    def methX(self):
        print("method of X")


class B(A):
    def __init__(self, a,b):
        super().__init__(a)
        self.b = b
    def methB(self):
        print("method of A")

class C(B):
    def __init__(self,a,b,c):
        super().__init__(a,b)
        self.c = c
    def methC(self):
        print("method of c")

# multiple inheritance

class D(A,X):
    def __init__(self, a,x,d):
        
        A.__init__(self,a)
        X.__init__(self,x)
        self.d = d
    def MethD(self):
        print("this is D's method")

obja = A('a')

objb = B('a','b')
objc = C('a','b','c')

print(objc.a)


objb.methA()

objd = D('a','x','d')

objd.MethD()
```

### Python Iterable vs Iterators
An iterator is an object that contains a countable number of values.

An iterator is an object that can be iterated upon, meaning that you can traverse through all the values.

Technically, in Python, an iterator is an object which implements the iterator protocol, which consist of the methods __iter__() and __next__().

Iterator vs Iterable
Lists, tuples, dictionaries, and sets are all iterable objects. They are iterable containers which you can get an iterator from.

All these objects have a iter() method which is used to get an iterator:



```python
lst = [12,3,4,5,6,7]

for i in lst:
    print(i)
```

##### Iterators are a subset of iterables whose values can not be accessed at the sametime, as they're not all stored in memory at once

```python
lst1 = iter(lst)

lst1
```

```python
next(lst1)


```

```python
my_list = [10, 20, 30]
my_iterator = iter(my_list) # Get the iterator

# 1. Get the first item
item1 = next(my_iterator) # Calls my_iterator.__next__()
print(f"First item: {item1}")

# 2. Get the second item
item2 = next(my_iterator)
print(f"Second item: {item2}")

# 3. Get the third item
item3 = next(my_iterator)
print(f"Third item: {item3}")

# 4. Try to get the fourth item
try:
    next(my_iterator)
except StopIteration:
    print("\nStopIteration raised! The loop would now terminate.")
```

The simplest way to think about this is:

An Iterable is something you can loop over (like a list or a string).

An Iterator is the object that remembers the state of the iteration (where it is in the loop) and gives you the next item.

## Generators

A generator is a function that, instead of returning a value once and terminating, yields a sequence of values over time.

Feature,return (Normal Function),yield (Generator Function)
Action:Return returns a value and terminates the function permanently :: Yield Pauses the function's execution and sends a value back to the caller.
State: Return => All local variables are destroyed. :: Yield => The local variables and execution state are saved until the next call.
Output, Return => A single value (or collection). Yield => A generator object (which is an iterator).

#### How Generators Work (Lazy Evaluation)
Generators implement the iterator protocol (__iter__ and __next__) automatically. Their power comes from lazy evaluation: they compute and output values one at a time, only when requested.

This makes them incredibly memory efficient because they don't store the entire sequence in memory at once. For example, generating a million numbers only requires memory for the current number, not the whole list.

```python
def simple_generator():
    print("Starting generator...")
    yield 1
    print("Paused and resumed...")
    yield 2
    print("Paused and resumed again...")
    yield 3
    print("Generator finished.")

# 1. Calling the function creates a generator object (an iterator)
gen = simple_generator() 
print(f"Generator Object: {gen}")

print("\n--- Manual Iteration ---")
# 2. Call next() to start/resume execution until the next yield
print(f"First output: {next(gen)}")
print(f"Second output: {next(gen)}")

print("\n--- Final Iteration ---")
# 3. The last call raises StopIteration automatically
print(f"Third output: {next(gen)}")

# Trying to call next() again would raise StopIteration
# try:
#     next(gen)
# except StopIteration:
#     print("Caught StopIteration")
```

```python
# List Comprehension: Creates and stores a list of 100 squares instantly
list_of_squares = [x*x for x in range(100)]
print(f"List type: {type(list_of_squares)}") # <class 'list'>

# Generator Expression: Creates a generator object that computes squares lazily
generator_of_squares = (x*x for x in range(100))
print(f"Generator type: {type(generator_of_squares)}") # <class 'generator'>

# We can iterate over the generator expression using a for loop (calling next() automatically)
print("\nFirst 5 squares from generator:")
for i, sq in enumerate(generator_of_squares):
    if i >= 5:
        break
    print(sq, end=" ")
```
