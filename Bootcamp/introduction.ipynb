##### Largest element in list

```python
#larges element in a list 

listmax = [2,4,44,5,66,7]

max_item = 0

for i in listmax:
    if i > max_item:
        max_item = i

print(f"Maximum item is {max_item}")

print(f"using inbuilt method: {max(listmax)}")
```

#### Swap two elements in a list

```python
my_list = [1,2,3,4,5]

initial_first = my_list[0]

my_list[0] = my_list[-1]
my_list[-1] = initial_first

print(my_list)


#alternatively: 

my_list[0], my_list[-1] = my_list[-1], my_list[0]

print(my_list)
```

```python
# swap the elements given

#3,4
print(my_list)
indx1 = my_list.index(3)
indx2 = my_list.index(4)

my_list[indx1], my_list[indx2] = my_list[indx2],my_list[indx1]

print(my_list)
```

### Reverse List

```python
print(f"original list: {my_list}")

list_len = len(my_list)-1

my_list_reversed = []

i = list_len
while list_len >= 0:
    item = my_list[list_len]
    my_list_reversed.append(item)

    list_len -= 1
print(f"List Reversed manually: {my_list_reversed}")
print(f"my_list:{my_list}")

auto_reverse = my_list.reverse()

#alternatively

print(f"List Reversed Automatically: {my_list[::-1]}")





```
